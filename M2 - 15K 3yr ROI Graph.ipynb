```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

# Matplotlib Theme
mpl.rcParams.update({
    'font.family': 'DejaVu Sans',
    'axes.titlesize': 18,
    'axes.labelsize': 15,
    'xtick.labelsize': 13,
    'ytick.labelsize': 13,
    'legend.fontsize': 13,
    'axes.facecolor': '#f5f5f2', 
    'figure.facecolor': '#f5f5f2',
    'axes.edgecolor': '#444444',
    'xtick.color': '#444444',
    'ytick.color': '#444444',
    'grid.color': '#bbbbbb',
    'axes.spines.top': False,
    'axes.spines.right': False
})

# Load and clean the data
df = pd.read_csv(r"C:\Users\murra\Downloads\M2_ROI_15KProb_2025-07-19(Sheet1).csv")
df = df.dropna(subset=["X1", "Y1"])
df["X1"] = pd.to_numeric(df["X1"], errors="coerce")
df["Y1"] = pd.to_numeric(df["Y1"], errors="coerce")
df = df.dropna(subset=["X1", "Y1"])
df = df.sort_values(by="X1")

# Define ROI Threshold
threshold_0 = 0
threshold_15k = 15000

# Plot setup
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(df["X1"], df["Y1"], color='#1f77b4', linewidth=3, label="Cumulative Probability")
ax.tick_params(axis='both', colors='black')

# Fill areas before and after threshold
ax.fill_between(df["X1"], 0, df["Y1"], where=(df["X1"] <= threshold_0), color='gray', alpha=0.2, label="ROI ≤ $0")
ax.fill_between(df["X1"], 0, df["Y1"], where=(df["X1"] > threshold_0), color='#1f77b4', alpha=0.2, label="ROI > $0")

# Vertical line at $15K
ax.axvline(threshold_15k, color='green', linestyle='--', linewidth=1.5)

# Text Annotations
ax.text(threshold_0 + 5000, 0.985, '21.1%', fontsize=15, ha='center', color='black', bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round,pad=0.3'))
ax.text(threshold_15k + 8000, 0.985, '79%', fontsize=15, ha='center', color='black', bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round,pad=0.3'))
ax.text(threshold_15k + 500, 0.5, "$15K ROI Threshold", verticalalignment='bottom', horizontalalignment='left', fontsize=13, color='green', rotation=0, bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round,pad=0.3'))

# Grid and Legend
ax.grid(True, linewidth=1, alpha=1)
legend = ax.legend(loc='lower right', frameon=True, facecolor='white', edgecolor='gray', framealpha=1, borderpad=1, labelspacing=0.8, bbox_to_anchor=(1, 0.05))
legend.get_frame().set_boxstyle('round,pad=0.4')

# Labels and title
ax.set_xlabel("ROI $")
ax.set_xlim(left=-5000) 
plt.tight_layout()
plt.show()
```
