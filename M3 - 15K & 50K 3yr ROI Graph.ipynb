```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

# Matplotlib Theme 
mpl.rcParams.update({
    'font.family': 'DejaVu Sans',
    'axes.titlesize': 18,
    'axes.labelsize': 15,
    'xtick.labelsize': 13,
    'ytick.labelsize': 13,
    'legend.fontsize': 13,
    'axes.facecolor': '#f5f5f2',
    'figure.facecolor': '#f5f5f2',
    'axes.edgecolor': '#444444',
    'xtick.color': '#444444',
    'ytick.color': '#444444',
    'grid.color': '#bbbbbb',
    'axes.spines.top': False,
    'axes.spines.right': False
})

# Load and Clean Data
df = pd.read_csv(r"C:\Users\murra\Downloads\M3_ROI_15KProb_2025-07-19(Sheet1).csv")  # Replace as needed
df = df.dropna(subset=["X1", "Y1"])
df["X1"] = pd.to_numeric(df["X1"], errors="coerce")
df["Y1"] = pd.to_numeric(df["Y1"], errors="coerce")
df = df.dropna(subset=["X1", "Y1"])
df = df.sort_values(by="X1")

# Threshold
threshold_0 = 0
threshold_custom = 15000  
label_low = "0%"        
label_high = "100%"      

# Plot Setup 
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(df["X1"], df["Y1"], color='#1f77b4', linewidth=3, label="Cumulative Probability")
ax.tick_params(axis='both', colors='black')

# Fill Areas
ax.fill_between(df["X1"], 0, df["Y1"], where=(df["X1"] <= threshold_0), color='gray', alpha=0.2, label="ROI ≤ $0")
ax.fill_between(df["X1"], 0, df["Y1"], where=(df["X1"] > threshold_0), color='#1f77b4', alpha=0.2, label="ROI > $0")

# Vertical Line (Customize threshold)
ax.axvline(threshold_custom, color='green', linestyle='--', linewidth=1.5)

# Text Annotations
ax.text(threshold_0 - 15000, 0.985, label_low, fontsize=15, ha='center', color='black', bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round,pad=0.3'))
ax.text(threshold_custom + 100000, 0.985, label_high, fontsize=15, ha='center', color='black', bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round,pad=0.3'))
ax.text(threshold_custom + 7500, 0.52, f"${int(threshold_custom/1000)}K ROI Threshold", verticalalignment='bottom', horizontalalignment='left', fontsize=13, color='green', rotation=0, bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round,pad=0.3'))

# Grid and Legend 
legend = ax.legend(loc='lower right', frameon=True, facecolor='white', edgecolor='gray',framealpha=1, borderpad=1, labelspacing=0.8, bbox_to_anchor=(1, 0.05))
legend.get_frame().set_boxstyle('round,pad=0.4')
ax.grid(True, linewidth=1, alpha=1)

# Labels and X-Axis Limit
ax.set_xlabel("ROI $")
ax.set_xlim(left=-50000) 
plt.tight_layout()
plt.show()
```
