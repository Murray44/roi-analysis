```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

# Matplotlib Theme
mpl.rcParams.update({
    'font.family': 'DejaVu Sans',
    'axes.titlesize': 18,
    'axes.labelsize': 15,
    'xtick.labelsize': 13,
    'ytick.labelsize': 13,
    'legend.fontsize': 13,
    'axes.facecolor': '#f5f5f2', 
    'figure.facecolor': '#f5f5f2',
    'axes.edgecolor': '#444444',
    'xtick.color': '#444444',
    'ytick.color': '#444444',
    'grid.color': '#bbbbbb',
    'axes.spines.top': False,
    'axes.spines.right': False
})

# ROI summary statistics
mean = 6.43
std_dev = 3.04
min_val = 0.75
max_val = 20.77

# Create x values (ROI range in $K)
x = np.linspace(min_val - 3, max_val + 3, 1000)
pdf = (1 / (std_dev * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / std_dev)**2)
cdf = np.cumsum(pdf) / np.sum(pdf)

# Plot
fig, ax = plt.subplots(figsize=(12, 7))
ax.plot(x, cdf, color='#1f77b4', linewidth=3, label="Cumulative Probability")
ax.tick_params(axis='both', colors='black')


# Highlight threshold lines
ax.axvline(0, color='black', linestyle='--', linewidth=1)
ax.axvline(15, color='green', linestyle='--', linewidth=1)

# Fill areas
ax.fill_between(x, 0, cdf, where=(x <= 0), color='gray', alpha=0.2, label="ROI ≤ $0 (0.5%)")
ax.fill_between(x, 0, cdf, where=(x > 0), color='#1f77b4', alpha=0.2, label="ROI > $0 (95.5%)")

# Text Annotations
ax.text(-1.3, 0.06, '$0 ROI', ha='right', fontsize=15, color='black')
ax.text(13.8, 0.67, '$15K ROI', ha='left', fontsize=15, color='green',bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round,pad=0.3'))
ax.text(-2.0, 0.985, '0.5%', fontsize=15, ha='center', color='black', bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round,pad=0.3'))
ax.text(8, 0.985, '95.5%', fontsize=15, ha='center', color='black', bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round,pad=0.3'))

# Title & Axis Labels
# ax.set_title("Cumulative Probability Distribution of 3-Year ROI", weight='bold')
ax.set_xlabel("ROI ($ Thousands)", labelpad=10)
ax.set_ylabel("Cumulative Probability", labelpad=10)

# Grid and Legend
ax.grid(True, linestyle='--', linewidth=0.7, alpha=0.5)
legend = ax.legend(
    loc='lower right',
    frameon=True,
    facecolor='white',
    edgecolor='gray',
    framealpha=1,
    borderpad=1,
    labelspacing=0.8,
    bbox_to_anchor=(1, 0.05)
)
legend.get_frame().set_boxstyle('round,pad=0.4')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl


# Style 
mpl.rcParams['font.family'] = 'DejaVu Sans'
mpl.rcParams['axes.titlesize'] = 16
mpl.rcParams['axes.labelsize'] = 14
mpl.rcParams['xtick.labelsize'] = 12
mpl.rcParams['ytick.labelsize'] = 12
mpl.rcParams['legend.fontsize'] = 12

# ROI summary statistics
mean = 6.43
std_dev = 3.04
min_val = 0.75
max_val = 20.77

# Create x values (ROI range in $K)
x = np.linspace(min_val - 3, max_val + 3, 1000)
pdf = (1 / (std_dev * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / std_dev)**2)
cdf = np.cumsum(pdf) / np.sum(pdf)

# Plot
fig, ax = plt.subplots(figsize=(12, 7))
ax.plot(x, cdf, color='#1f77b4', linewidth=3, label="Cumulative Probability")
xticks = ax.get_xticks()
ax.set_xticklabels([str(int(tick * 10)) for tick in xticks])
ax.tick_params(axis='both', colors='black')


# Highlight zones
ax.axvline(0, color='black', linestyle='--', linewidth=1)
ax.axvline(5.0, color='green', linestyle='--', linewidth=1) 

# Fill areas under the curve
ax.fill_between(x, 0, cdf, where=(x <= 0), color='gray', alpha=0.15, label="Shaded area: ROI ≤ $0 (10.3%)")
ax.fill_between(x, 0, cdf, where=(x >= 0), color='#1f77b4', alpha=0.15, label="Shaded area: ROI > $0 (89.7%)")


# Text annotations (updated to reflect new stats)
ax.text(-0.4, 0.05, '$0 ROI', ha='right', fontsize=15, color='black')
ax.text(4.9, 0.64, '$50K ROI', ha='center', fontsize=15, color='green', bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round,pad=0.3'))
ax.text(-2.0, 0.985, '10.3%', fontsize=15, ha='center', color='black', bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round,pad=0.3'))
ax.text(8.5, 0.985, '89.7%', fontsize=15, ha='center', color='black', bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round,pad=0.3'))

# Title and axis labels
# ax.set_title("Cumulative Probability Distribution of 3-Year ROI")
ax.set_xlabel("ROI ($ Tens of Thousands)")
ax.set_ylabel("Cumulative Probability")

# Grid and legend
ax.grid(True, linestyle='--', linewidth=0.7, alpha=0.5)
legend = ax.legend(
    loc='lower right',
    frameon=True,
    facecolor='white',
    edgecolor='gray',
    framealpha=1,
    borderpad=1,
    labelspacing=0.8,
    bbox_to_anchor=(1, 0.05)
)
legend.get_frame().set_boxstyle('round,pad=0.4')

plt.tight_layout()
plt.show()
```
